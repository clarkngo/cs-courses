```python
%pip install nltk textblob
import nltk
from textblob import TextBlob
import random

# Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')

# Core dataset: question-answer pairs
dataset = {
    "how are you": "I'm just a bunch of code, but I'm doing great!",
    "what is your name": "I'm ChatBot, your friendly assistant.",
    "tell me a joke": "Why did the computer show up at work late? It had a hard drive!",
    "what is the capital of France": "The capital of France is Paris.",
    "who created you": "I was created by a team of developers.",
    "what can you do": "I can answer questions, tell jokes, and analyze sentiment.",
    "are you intelligent": "I try my best, but I'm still learning!",
    "do you like humans": "I think humans are fascinating!",
    "what is the meaning of life": "42, according to some famous sources.",
    "can you help me": "Of course! Ask me anything."
}

# Multiple response variations (bonus)
response_variations = {
    "how are you": [
        "I'm just a bunch of code, but I'm doing great!",
        "I'm functioning as expected, thank you!",
        "All systems operational!"
    ],
    "tell me a joke": [
        "Why did the computer show up at work late? It had a hard drive!",
        "Why do programmers prefer dark mode? Because light attracts bugs!",
        "Why was the math book sad? It had too many problems."
    ]
}

# Handle user input: exact match
def handle_user_input(user_input):
    user_input_norm = user_input.lower().strip()
    if not user_input_norm:
        return "Please enter something for me to respond to."
    for question in dataset:
        if user_input_norm == question:
            # Support multiple response variations
            if question in response_variations:
                return random.choice(response_variations[question])
            return dataset[question]
    return handle_ambiguous_input(user_input_norm)

# Handle ambiguous input: partial match
def handle_ambiguous_input(user_input):
    user_tokens = set(nltk.word_tokenize(user_input))
    best_match = None
    max_overlap = 0
    for question in dataset:
        question_tokens = set(nltk.word_tokenize(question))
        overlap = len(user_tokens & question_tokens)
        if overlap > max_overlap:
            max_overlap = overlap
            best_match = question
    if best_match and max_overlap > 0:
        if best_match in response_variations:
            return random.choice(response_variations[best_match])
        return dataset[best_match]
    return "Sorry, I didn't quite understand that. Can you rephrase?"

# Sentiment analysis
def analyze_sentiment(text):
    try:
        blob = TextBlob(text)
        polarity = blob.sentiment.polarity
        subjectivity = blob.sentiment.subjectivity
        return f"Sentiment polarity: {polarity:.2f}, subjectivity: {subjectivity:.2f}"
    except Exception:
        return "Sentiment analysis failed."

# Enhanced user input handler: combines response and sentiment
def enhanced_handle_user_input(user_input):
    response = handle_user_input(user_input)
    sentiment = analyze_sentiment(user_input)
    return f"{response}\n{sentiment}"

# Main chat loop
def start_chat():
    print("Hello! I'm ChatBot. Type 'exit', 'quit', or 'bye' to end the conversation.")
    while True:
        user_input = input("You: ")
        if user_input.lower().strip() in ["exit", "quit", "bye"]:
            print("ChatBot: Goodbye! Have a great day.")
            break
        print("ChatBot:", enhanced_handle_user_input(user_input))

# Test new questions
def test_dataset():
    for question in dataset:
        response = handle_user_input(question)
        assert response is not None and response != "", f"No response for: {question}"

# Test sentiment analysis
def test_sentiment():
    test_sentences = [
        "I love sunny days!",
        "This is terrible.",
        "I'm not sure how I feel.",
        "",
        "!!!"
    ]
    for sentence in test_sentences:
        print(f"Input: {sentence}")
        print(analyze_sentiment(sentence))

# Edge case testing
def test_edge_cases():
    edge_inputs = ["", "   ", "!!!", "12345", "unknown question"]
    for inp in edge_inputs:
        print(f"Input: '{inp}'")
        print(handle_user_input(inp))

test_dataset()
test_sentiment()
test_edge_cases()
start_chat()
```
