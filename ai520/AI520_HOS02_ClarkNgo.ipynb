```python
text = """Natural Language Processing is fascinating. It helps computers understand human language. This technology is used everywhere, from virtual assistants to translation services."""
```

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(text)
print("Tokenized words:")
print(words)
```

```python
from nltk import FreqDist

fdist = FreqDist(words)
print("Most common words:")
print(fdist.most_common(10))
```

```python
import string
nltk.download('stopwords')
from nltk.corpus import stopwords

# Remove punctuation
text_no_punct = "".join([char for char in text if char not in string.punctuation])

# Tokenize and remove stopwords
stop_words = set(stopwords.words('english'))
words_no_stop = [word.lower() for word in word_tokenize(text_no_punct) if word.lower() not in stop_words]

print("Clean text tokens:")
print(words_no_stop)
```

```python
fdist_clean = FreqDist(words_no_stop)
print("Most common words after cleaning:")
print(fdist_clean.most_common(10))
```

```python
import matplotlib.pyplot as plt

fdist_clean.plot(30, cumulative=False)
plt.title('Word Frequency Distribution')
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.show()
```
