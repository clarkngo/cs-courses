```python
import random

def number_game():
    num = random.randint(20, 25)
    print("Try to guess your number:")
    print("-" * 50)

    while True:
        try:
            guess = int(input("Pick a number between 20 and 25: "))
            if guess == num:
                print(f"Perfect, you did it! {guess}")
                break
            print(f"Sorry. Try again! {guess}")
        except ValueError:
            print("Please enter a valid number")

if __name__ == "__main__":
    number_game()
```

```python
# List comprehension for text processing
words = ["Natural", "Language", "Processing"]
word_lengths = [len(word) for word in words]

# Using f-strings for formatting
print(f"Words: {words}")
print(f"Lengths: {word_lengths}")

# Dictionary comprehension for word counting
text = "Natural Language Processing is fascinating"
word_count = {word: text.count(word) for word in text.split()}
print(f"Word frequencies: {word_count}")
```

```python
import torch
from transformers import pipeline

# Using modern NLP pipeline
classifier = pipeline("sentiment-analysis")
result = classifier(text)
print("\nSentiment Analysis:")
print(result)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("\nNamed Entities:")
for ent in doc.ents:
    print(f"- {ent.text} ({ent.label_})")

print("\nPart-of-Speech Tags:")
for token in doc[:5]:
    print(f"- {token.text}: {token.pos_}")
```
