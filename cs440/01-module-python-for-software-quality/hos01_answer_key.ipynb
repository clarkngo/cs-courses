### Exercise 1

### Exercise 2
**Issue:** The `=` operator is used for assignment, while `==` is used for comparison. The code should compare the `name` variable with an empty string using `==`.

```
def greet(name):
    if name == "":
        return "Hello, Guest!"
    return f"Hello, {name}"

print(greet(""))
```

### Exercise 3
**Issue:** The variable `radus` is a typo and should be corrected to `radius` to match the parameter name and avoid a NameError.

```
def circle_area(radius):
    return 3.14 * radius * radius

print(circle_area(5))
```

### Exercise 4
**Issue:** The loop is incorrectly starting from index 1, causing the first element to be missed in the summation. It should start from index 0.

```
def total(lst):
    sum = 0
    for i in range(len(lst)):
        sum += lst[i]
    return sum

print(total([1, 2, 3, 4]))
```

### Exercise 5
**Issue:** The code attempts to divide by zero, which is not allowed. We should check if `b` is zero before performing the division to avoid a runtime error.

```
def divide(a, b):
    if b == 0:
        return 'Cannot divide by zero'
    return a / b

print(divide(10, 0))
```
