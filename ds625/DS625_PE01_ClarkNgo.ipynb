```python
# 1. Create a list of integers from 1 to 99
# Note: range(1, 100) stops at 99. Use range(1, 101) to include 100.
numbers = list(range(1, 100))
```

```python
# 2. Generate a list of squared numbers
# Logic: We take each 'x' and square it (x**2) for the new list
squared_numbers = [x ** 2 for x in numbers]
```

```python
# 3. Calculate the cumulative total of the squared values
total_sum_of_squared = sum(squared_numbers)
```

```python

# 4. Display the final result using an f-string
print(f"The sum of squared numbers between 1 and 100 is: {total_sum_of_squared}")
```
