```python
from pyspark.sql import SparkSession

# Initialize Spark Session
spark = SparkSession.builder.appName("SalesDataLab").getOrCreate()

# Create DataFrame
data = [("Jane", 1287), ("Joe", 2221), ("Jane", 3000), 
        ("Peter", 1798), ("Jane", 1990), ("Joe", 2339), 
        ("Jane", 2067), ("Joe", 2470), ("Peter", 1072), 
        ("Jane", 2477)]

columns = ["Customer", "Sales"]
sales_df = spark.createDataFrame(data, columns)

# Register DataFrame as a SQL view
sales_df.createOrReplaceTempView("sales_tbl")
```

```python
# Register DataFrame as a SQL view
sales_df.createOrReplaceTempView("sales_tbl")

# Select all data using SQL
spark.sql("SELECT * FROM sales_tbl").show()
```

```python
# Use SQL to create a new column dividing sales by 1000
spark.sql("SELECT Customer, Sales, Sales / 1000 AS Sales_in_Thousands FROM sales_tbl").show()
```

```python
# Use SQL to filter for customer 'Jane'
spark.sql("SELECT * FROM sales_tbl WHERE Customer = 'Jane'").show()
```

```python
# Use SQL to filter for customer 'Jane'
spark.sql(" SELECT Customer, AVG(Sales) AS Avg_Sales FROM sales_tbl  GROUP BY Customer ").show()
```

```python
# Use SQL to calculate total sales per customer
spark.sql("SELECT Customer, SUM(Sales) AS Total_Sales FROM sales_tbl GROUP BY Customer").show()
```

```python
# Use SQL to find maximum and minimum sales values
spark.sql("SELECT MAX(Sales) AS Max_Sales, MIN(Sales) AS Min_Sales FROM sales_tbl").show()
```
