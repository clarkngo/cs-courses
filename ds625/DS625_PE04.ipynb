```python
# Create employee data
data = [
    ('Alice', 'Engineering', 75000, 5),
    ('Bob', 'Marketing', 60000, 3),
    ('Charlie', 'Engineering', 95000, 10),
    ('Diana', 'HR', 50000, 4),
    ('Eve', 'Marketing', 80000, 8),
    ('Frank', 'Engineering', 70000, 6),
    ('Grace', 'HR', 55000, 5)
]

# Define schema
columns = ['Employee', 'Department', 'Salary', 'Years_Experience']

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Register as temporary SQL view
df.createOrReplaceTempView('employees')

# Display the data
display(df)
```

```python
%sql
SELECT 
  Employee,
  Department,
  Salary,
  Years_Experience,
  ROUND(Salary / Years_Experience, 2) AS Salary_Per_Year
FROM employees
```

```python
%sql
SELECT 
  Employee,
  Department,
  Salary,
  Years_Experience,
  ROUND(Salary / Years_Experience, 2) AS Salary_Per_Year
FROM employees
WHERE (Salary / Years_Experience) > 8000
```

```python
%sql
SELECT 
  Department,
  SUM(Salary) AS Total_Salary,
  COUNT(*) AS Employee_Count,
  ROUND(AVG(Salary), 2) AS Avg_Salary
FROM employees
GROUP BY Department
ORDER BY Total_Salary DESC
```

```python
%sql
SELECT 
  Department,
  SUM(Salary) AS Total_Salary
FROM employees
GROUP BY Department
ORDER BY Total_Salary DESC
```
