```python
from pyspark.sql import SparkSession

# Initialize Spark Session
spark = SparkSession.builder.appName("PySparkDataFrameLab").getOrCreate()
```

```python
# Load JSON data
students = spark.table("students"). select ("id", "first", "last", "grade") # Use the table name you provided
students.show()
```

```python
# Filter students with grade > 3.6
filtered_students = students.filter(students.grade > 3.6)
filtered_students.show()
```

```python
# Select columns and order by grade
sorted_students = students.select("first", "last", "grade").orderBy("grade", ascending=False)
sorted_students.show()
```

```python
# Group by last name and calculate average grade
avg_grade = students.groupBy("last").avg("grade")
avg_grade.show()
```

```python
# Write the filtered DataFrame to a temporary table
filtered_students.write.mode("overwrite").saveAsTable("default.temp_students")

# Read the materialized table and display its contents
filtered_students.show()
```

```python
# Drop the temporary table to clean up resources after use
spark.sql("DROP TABLE IF EXISTS default.temp_students")
```
