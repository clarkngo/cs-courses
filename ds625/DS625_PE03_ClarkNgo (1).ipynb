```python
# Step 1: Create a DataFrame for product sales (Product, Sales, Quantity)
from pyspark.sql.functions import col, sum as spark_sum

# Sample product sales data
data = [
    ("Laptop", 1200, 8),
    ("Phone", 800, 15),
    ("Tablet", 500, 25),
    ("Laptop", 1200, 10),
    ("Monitor", 300, 40),
    ("Phone", 800, 12),
    ("Keyboard", 50, 100),
    ("Laptop", 1200, 5)
]

columns = ["Product", "Sales", "Quantity"]
df = spark.createDataFrame(data, columns)

print("Step 1: Original DataFrame")
display(df)
```

```python
# Step 2: Add a new column with the total sales value using withColumn()
df_with_total = df.withColumn("Total_Sales", col("Sales") * col("Quantity"))

print("\nStep 2: DataFrame with Total_Sales column")
display(df_with_total)
```

```python
# Step 3: Filter the DataFrame for transactions with total sales greater than $10,000
df_filtered = df_with_total.filter(col("Total_Sales") > 10000)

print("\nStep 3: Filtered DataFrame (Total_Sales > $10,000)")
display(df_filtered)
```

```python
# Step 4: Group the data by product and calculate the total sales using groupBy() and agg()
df_grouped = df_filtered.groupBy("Product").agg(
    spark_sum("Total_Sales").alias("Total_Sales")
)

print("\nStep 4: Grouped by Product with aggregated Total_Sales")
display(df_grouped)
```

```python
# Step 5: Select the Product and Total_Sales columns and display the result
df_final = df_grouped.select("Product", "Total_Sales")

print("\nStep 5: Final Result (Product and Total_Sales)")
display(df_final)
```
