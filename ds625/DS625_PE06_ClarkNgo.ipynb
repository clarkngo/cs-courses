```python
from pyspark.sql import SparkSession

# Initialize Spark Session
spark = SparkSession.builder.appName("PySparkDataFrameLab").getOrCreate()
```

```python
# Load JSON data
employees = spark.table("employees").select ("id", "first_name", "last_name", "department", "salary", "years_experience") # Use the table name you provided
employees.show()
```

```python
# Filter employees with salary > 80,000
filtered_employees = employees.select ("first_name", "last_name", "department", "salary").filter(employees.salary > 80000)
filtered_employees.show()
```

```python
# Select columns and order by years of experience
sorted_employees = employees.select ("first_name", "last_name", "years_experience", "department").orderBy("years_experience", ascending=False)
sorted_employees.show()
```

```python
# Group by last name and calculate average salary
avg_salary = employees.groupBy("department").avg("salary")
avg_salary.show()
```

```python
# Write the filtered DataFrame to a temporary table
filtered_employees.write.mode("overwrite").saveAsTable("default.temp_employees")

# Read the materialized table and display its contents
filtered_employees.show()
```

```python
# Drop the temporary table to clean up resources after use
spark.sql("DROP TABLE IF EXISTS default.temp_employees")
```
