```python
# 1. Create the vectors x = [1, 2, 3, 4, 5] and y = [2, 5, 6, 7, 10]. Remember to reshape x by using reshape(-1, 1).

import numpy as np

x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 5, 6, 7, 10])
print("x:\n", x)
print("y:", y)
```

```python
# 2. Calculate the values of m and b in the equation y = m * x + b (for the x and y defined in part 1). Print the result.
from sklearn.linear_model import LinearRegression

fit = LinearRegression().fit(x, y)
m = fit.coef_[0]
b = fit.intercept_
print("m:", m)
print("b:", b)
```

```python
# 3. Plot y vs x and the regression line.
import matplotlib.pyplot as plt

plt.plot(x, y, 'o')
plt.plot(x, m * x + b)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# 4. Calculate and print the Sum of Square Errors (the sum of the squares of the residuals)
y_actual = y
y_predict = fit.predict(x)
sse = np.sum((y_actual - y_predict) ** 2)
print("Sum of the squares of residuals:", sse)
```

```python
# 5. Plot the residual vs x.
residuals = y_actual - y_predict

plt.plot(x, residuals, 'o')
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel("x")
plt.ylabel("Residuals")
plt.show()
```
