```python
# Plot the function f = x^2 + 10
from sympy import *
x = symbols('x')
f = x**2 + 10
plot(f)
```

```python
from sympy import *
from sympy.plotting import plot3d
x, y = symbols('x y')
f = x**2 + y**2
plot3d(f)
```

```python
# Calculate the factorial of 10
n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("The factorial of", n, "is : ", end="")
print(fact)
```

```python
# Calculate 3^3
base = 3
exponent = 3
result = pow(base, exponent)
print("Answer = " + str(result))
```

```python
# Calculate the log in base 3 of 81
from math import log

x = log(81, 3)

print(x)
```
