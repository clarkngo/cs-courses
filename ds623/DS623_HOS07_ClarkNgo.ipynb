```python
# 1. Defining Vectors x and y
import numpy as np

x = np.array([6, 7, 8, 9, 10])
x = x.reshape(-1, 1)
y = np.array([5, 10, 10, 15, 14])
print("x:", x)
print("y:", y)
```

```python
# 2. Fitting a Linear Regression Model (m and b)
import numpy as np
from sklearn.linear_model import LinearRegression

x = np.array([6, 7, 8, 9, 10]).reshape(-1, 1)
y = np.array([5, 10, 10, 15, 14])
fit = LinearRegression().fit(x, y)
m = fit.coef_.flatten()
b = fit.intercept_.flatten()
print("m:", m)
print("b:", b)
```

```python
# 3. Plotting the Scatterplot and Regression Line
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

x = np.array([6, 7, 8, 9, 10]).reshape(-1, 1)
y = np.array([5, 10, 10, 15, 14])
fit = LinearRegression().fit(x, y)
m = fit.coef_[0]
b = fit.intercept_
plt.plot(x, y, 'o')  # scatterplot
plt.plot(x, m * x + b)  # line
plt.show()
```

```python
# 4. Calculating Sum of Squares of Residuals (SSE)
import numpy as np
from sklearn.linear_model import LinearRegression

x = np.array([6, 7, 8, 9, 10]).reshape(-1, 1)
y = np.array([5, 10, 10, 15, 14])
fit = LinearRegression().fit(x, y)
y_actual = y
y_predict = fit.predict(x)
sse = np.sum((y_actual - y_predict) ** 2)
print("Sum of the squares of residuals:", sse)
```
