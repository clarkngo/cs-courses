```python
# Section 1: Working with vectors
# Declaring a vector using basic Python:
v = [3, 2]
print(v)
```

```python
# Defining a vector using NumPy:
import numpy as np
v = np.array([3, 2])
print(v)
```

```python
# Defining two vectors using NumPy:
from numpy import array
v = array([3, 2])
w = array([2, -1])

print(v)
print(w)
```

```python
# Section 2: Adding and Combining Vectors
# Adding two vectors together:
from numpy import array
v = array([3, 2])
w = array([2, -1])

v_plus_w = v + w
print(v_plus_w)
```

```python
# Section 3: Scaling vectors
# Scaling a vector by multiplying it by a scalar:
from numpy import array
v = array([3, 1])

scaled_v = 2.0 * v
print(scaled_v)
```

```python
# Section 4: Matrix vector multiplication
# Defining a basis matrix and transforming a vector using the dot product:
from numpy import array
basis = array([
    [3, 0],
    [0, 2]
])
v = array([1, 1])
new_v = basis.dot(v)
print(new_v)
```

```python
# Transforming a vector using i_hat and j_hat vectors:
from numpy import array
i_hat = array([2, 0])
j_hat = array([0, 3])

basis = array([i_hat, j_hat]).transpose()
v = array([2, 1])

new_v = basis.dot(v)
print(new_v)
```

```python
# Combining two transformations:
from numpy import array
i_hat1 = array([0, 1])
j_hat1 = array([-1, 0])
transform1 = array([i_hat1, j_hat1]).transpose()

i_hat2 = array([1, 0])
j_hat2 = array([1, 1])
transform2 = array([i_hat2, j_hat2]).transpose()

combined = transform2 @ transform1
v = array([1, 2])
print(combined.dot(v))
```
