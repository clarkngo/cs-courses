```python
# 1. Create the matrix M with the values [[2, 1], [4, 3]].
import numpy as np

M = np.array([[2, 1], [4, 3]])
print("Matrix M:\n", M)
```

```python
# 2. Calculate the determinant of the matrix M.
det_M = np.linalg.det(M)
print("Determinant of M:", det_M)
```

```python
# 3. Calculate inv_M as the inverse of the matrix M. Print inv_M.
inv_M = np.linalg.inv(M)
print("inv_M:\n", inv_M)
```

```python
# 4. What is the value of M * inv_M?
product_M = np.dot(M, inv_M) # or M @ inv_M
print("M * inv_M:\n", product_M)
```

```python
# 5. Use a matrix to solve the following system of equations. Name the matrix so that the equation is A X = b. Print the result.
# x + y + z = 3
# 2x + 3y + 4z = 9
# x - 2y + z = 0

A = np.array([[1, 1, 1],
              [2, 3, 4],
              [1, -2, 1]])
b = np.array([3, 9, 0])

X = np.linalg.solve(A, b)
print("Solution X (x, y, z):", X)
```

```python
# 6. Let's call A = [[1, 1, 1], [2, 3, 4], [1, -2, 1]]. Calculate the inverse of A and call it inv_A. Print the result.
inv_A = np.linalg.inv(A)
print("inv_A:\n", inv_A)
```

```python
# 7. Calculate inv_A * b. b is given in #5.
result_inv_A_b = np.dot(inv_A, b)
print("inv_A * b:\n", result_inv_A_b)
```

```python
# 8. Create the matrix A1 as [[5, 2, -4], [-10, 14, -8], [-5, 2, 6]]. Calculate and print the eigenvalues (lambda1, lambda2, & lambda3)
# and eigenvectors of A1 (v1, v2, & v3).
A1 = np.array([[5, 2, -4],
               [-10, 14, -8],
               [-5, 2, 6]])

eigenvalues, eigenvectors = np.linalg.eig(A1)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
# 9. Verify that lambda_i and v_i (i = 1, 2, 3) are correct eigenvalues and eigenvectors using the following relations:
# (A1 - lambda_i * v_i) = 0 for i = 1, 2, and 3.
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v = eigenvectors[:, i]
    # (A1 - lambda_i * I) * v_i should be close to 0 vector
    verification = np.dot((A1 - lam * np.identity(3)), v)
    print(f"Verification for lambda_{i+1}:", verification)
```
