```python
# 5. Create a plot of the function f = x**2 + x + 1

from sympy import *
x = symbols('x')
f = x**2 + x + 1
plot(f)
```

```python
# 6. Create a plot of the function f = x^2 - y^2

from sympy import *
from sympy.plotting import plot3d
x, y = symbols('x y')
f = x**2 - y**2
plot3d(f)
```
