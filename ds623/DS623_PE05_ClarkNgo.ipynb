```python
# 1. Define the numpy vectors v = [1, 1] and w = [2, 2].
import numpy as np

v = np.array([1, 1])
w = np.array([2, 2])
```

```python
# 2. Create a vector z that is the sum of v and w.
# Print z.
z = v + w
print(z)
```

```python
# 3. Create a vector v1 that is the result of scaling v by multiplying it by 3.
v1 = v * 3
```

```python
# 4. Create the matrix m1 with the values [[2, 0], [0, 3]]. Apply the transformation defined by m1 to v, save the result as the vector v2.
m1 = np.array([[2, 0], [0, 3]])
v2 = np.dot(m1, v) # or m1 @ v
```

```python
# 5. Create the matrix m2 with the values [[0, -1], [1, 0]].
m2 = np.array([[0, -1], [1, 0]])
```

```python
# 6. Create the matrix m3 that is the result of combining the transformations represented by m1 and m2.
m3 = np.dot(m2, m1) # or m2 @ m1
```

```python
#7. Apply the transformation represented by m3 to v to create the vector v3. Print v3.
v3 = np.dot(m3, v) # or m3 @ v
print(v3)
```

```python
#8. Two vectors in two different directions are linearly independent and have this unlimited span.
# In other words, it can create any vector in the same dimension.
# Let's define the numpy vector z = [1, 0].
# Suppose coefficients C1 and C2 are multiplied to vectors v and z to create a vector [2, 4].
# Calculate C1 and C2 (that makes C1*v + C2*z = [2, 4])
import numpy as np

# Ensure v and z have the correct values for step 8
v = np.array([1, 1])
z = np.array([1, 0])

target1 = np.array([2, 4])
A1 = np.column_stack((v, z))

C1, C2 = np.linalg.solve(A1, target1)
print(f"C1: {C1}, C2: {C2}")
```

```python
#9. Calculate C1 and C2 that create a vector [20, -4].
target2 = np.array([20, -4])
C1, C2 = np.linalg.solve(A1, target2)
print(f"C1: {C1}, C2: {C2}")
```

```python
# 10. This time, you want to use v1 = [2, 0] and z to span the two-dimensional space.
# Calculate C1 and C2 that create a vector [20, -4].
import numpy as np

v1_new = np.array([2, 0])
z = np.array([1, 0])
target2 = np.array([20, -4])

A2 = np.column_stack((v1_new, z))

try:
    C1_new, C2_new = np.linalg.solve(A2, target2)
    print(f"C1: {C1_new}, C2: {C2_new}")
except np.linalg.LinAlgError:
    print("No solution exists: matrix A2 is singular because v1 and z are linearly dependent.")
```
