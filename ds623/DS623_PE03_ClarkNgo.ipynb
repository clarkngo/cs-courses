```python
# This data will be used throught the questions

data = [1,  3,  2,  5,  7,  0,  2,  3]
weights = [.1, .2, .2, .1, .05, .05, .1, .2]
```

```python
# 1. Calculate the mean of the data
import numpy as np

mean_val = np.mean(data)
mean_val
```

```python
# 2. Calculate the weighted mean of the data (use data and weights)
weighted_mean = np.average(data, weights=weights)
weighted_mean
```

```python
# 3. Calculate the median of the data
median_val = np.median(data)
median_val
```

```python
# 4. Calculate the mode of the data
from scipy import stats

mode_val = stats.mode(data, keepdims=True).mode[0]
mode_val
```

```python
# 5. Calculate the variance of the data
# Using ddof=0 for population variance (standard default for numpy)
variance_val = np.var(data, ddof=0)
variance_val
```

```python
# 6. Calculate the standard deviation of the data
std_val = np.std(data, ddof=0)
std_val
```

```python
# Suppose a regional transit authority tracks weekday morning bus commutes along an urban corridor, finding travel times follow a normal distribution with an average duration of 35 minutes and a standard deviation of 10 minutes due to regular traffic and passenger boarding variations.
```

```python
# 7. Calculate the probability that a randomly selected morning bus commute takes less than 40 minutes.
mu = 35
sigma = 10
prob_less_than_40 = stats.norm.cdf(40, loc=mu, scale=sigma)
prob_less_than_40
```

```python
# 8. Calculate the percentage of bus trips on this corridor are expected to take between 30 and 40 minutes.
prob_between_30_40 = stats.norm.cdf(40, loc=mu, scale=sigma) - stats.norm.cdf(30, loc=mu, scale=sigma)
percentage_between_30_40 = prob_between_30_40 * 100
percentage_between_30_40
```

```python
# 9. Calculate the z-score of a bus commute that takes 50 minutes.
x_val = 50
z_score = (x_val - mu) / sigma
z_score
```
