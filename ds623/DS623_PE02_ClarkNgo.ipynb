```python
# 1. Calculate the probability that the roll of a die will produce a 3
num_sides = 6
three = 1
roll_three = three / num_sides
print("Probability of rolling three:", roll_three)
```

```python
# 2. Calculate the probability that two simultaneus rolls of two dice will produce
# the numbers 3 and 3
num_sides = 6
three = 1
roll_three = three / num_sides
roll_three_and_three = roll_three * roll_three
print("Probability of rolling three and three:", roll_three_and_three)
```

```python
# 3. Calculate the probability that two consecutive rolls of a die will produce
# the numbers 1 and 2
num_sides = 6
one = 1
two = 1
roll_one = one / num_sides
roll_two = two / num_sides
roll_one_and_two = roll_one * roll_two
print("Probability of rolling one and two:", roll_one_and_two)
```

```python
# 4. Calculate the probability that two simultaneous rolls of two dice will
# produce the numbers 3 or 4
num_sides = 6
three = 1
four = 1
roll_three_or_four = (three + four) / num_sides
print("Probability of rolling 3 or 4:", roll_three_or_four)
```

```python
# 5. Calculate the probability that two simultaneous rolls of two dice will
# produce 2 numbers that add up to 8.
pairs_of_eight = 5  # Pairs: (2,6), (3,5), (4,4), (5,3), (6,2)
possible_outcomes = 36
roll_total_eight = pairs_of_eight / possible_outcomes
print("Probability of rolling sum of 8:", roll_total_eight)
Python
```

```python
# 6. Calculate the probabiliy that it's raining given that you are carrying
# an umbrella given that you know the following probabilities
# (use Baye's theorem)

# Probability of bringing an umbrella = 0.70
# Probability of rain = 0.80
# Probability of bringing an umbrella given that it's raining = 0.85

p_umbrella = 0.70
p_rain = 0.80
p_umbrella_given_rain = 0.85

p_rain_given_umbrella = (p_umbrella_given_rain * p_rain) / p_umbrella
print("Probability of rain given umbrella:", p_rain_given_umbrella)
```

- Part 3: For Part 3, these are back-to-back rolls that don't affect each other, so they're independent events. I used the multiplication rule and just multiplied their individual probabilities together.   

- Part 4: For Part 4, I wanted to find the chance of rolling a 3 or a 4. Since you can't roll both at the exact same time on a single die, I used the addition rule to add their individual probabilities together.

- Part 6: For Part 6, I used Bayes' Theorem because we had a conditional probability and needed to flip it around. It's basically a way to update the chance of it raining once you know you're carrying an umbrella.
