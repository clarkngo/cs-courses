```python
# Section 1: Calculate the Mean of a data set
#Calculate the mean of a data set
sample = [1, 5, 2, 6, 7, 7, 3, 2]

mean = sum(sample) / len(sample)
print(mean)
```

```python
# Section 2: Calculate the Weighted Mean of a data set
#Calculate the weighted mean of a data set
sample = [70, 40, 56, 51]
weights = [.10, .30, .50, .40]

weighted_mean = sum(s * w for s,w in zip(sample, weights)) / sum(weights)

print(weighted_mean)
```

```python
# Section 3: Calculate the Median of a data set
#Calculate the median of a data set
sample = [2, 1, 3, 5, 7, 50, 24]

def median(values):
    ordered = sorted(values)
    print(ordered)
    n = len(ordered)
    mid = int(n / 2) - 1 if n % 2 == 0 else int(n/2)

    if n % 2 == 0:
        return (ordered[mid] + ordered[mid+1]) / 2.0
    else:
        return ordered[mid]

print(median(sample))
```

```python
# Section 4: Calculate the Mode of a data set
#Calculate the mode of a data set
from collections import defaultdict

sample = [5, 6, 3, 6, 5, 8, 2, 4]

def mode(values):
    counts = defaultdict(lambda: 0)

    for s in values:
        counts[s] += 1

    max_count = max(counts.values())
    modes = [v for v in set(values) if counts[v] == max_count]
    return modes

print(mode(sample))
```

```python
# Section 5: Calculate the Variance of a data set
#Calculate the variance of a data set
data = [5, 3, 6, 7, 8, 12, 15]

def variance(values):
    mean = sum(values) / len(values)
    _variance = sum((v - mean) ** 2 for v in values) / len(values)
    return _variance

print(variance(data))
```

```python
# Section 6: Calculate the Standard Deviation of a data set
#Calculate the standard deviation of a data set
from math import sqrt

data = [5, 6, 8, 3, 7, 16, 18]

def variance(values):
    mean = sum(values) / len(values)
    _variance = sum((v - mean) ** 2 for v in values) / len(values)
    return _variance

def std_dev(values):
    return sqrt(variance(values))

print(std_dev(data))
```

```python
# Section 7: Calculate the Probability of a Value Below a Given Score
from scipy.stats import norm

# Exam score distribution parameters
mu = 500
sigma = 50

# Calculate the probability of scoring below 550
p_below_550 = norm.cdf(550, loc=mu, scale=sigma)

print(f"P(score < 550): {p_below_550:.4f}")
```

```python
# Section 8: Calculate the Probability of Scoring Between Two Values
# Calculate the probability of scoring between 450 and 550
p_middle = norm.cdf(550, loc=mu, scale=sigma) - norm.cdf(450, loc=mu, scale=sigma)
print(f"P(450 < score < 550): {p_middle:.4f}")
```

```python
# Section 9: Calculate the Score at a Given Percentile
# Calculate the score at the 90th percentile
cutoff_top_10_percent = norm.ppf(0.90, loc=mu, scale=sigma)

print(f"90th percentile score: {cutoff_top_10_percent:.2f}")
```

```python
# Section 10: Calculate the Z-Score of a Given Value
# Student's exam score
student_score = 620

# Calculate the Z-score
z_score = (student_score - mu) / sigma

print(f"Z-score for a score of 620: {z_score:.2f}")
```
