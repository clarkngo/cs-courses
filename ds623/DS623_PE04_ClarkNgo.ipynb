```python
# 1. Calculate the z-score given the following parameters:
#   Distribution: Normal
#   Mean: 1
#   Standard Deviation: 1
#   Probability = 95%

import scipy.stats as stats

z_score = stats.norm.ppf(0.975, loc=1, scale=1)
print(f"Z-score for 95% probability (two-tailed): {z_score:.4f}")
```

```python
# 2. Calculate a confidence interval given the following parameters:
#   Distribution: Normal
#   Sample Mean: 65
#   Sample Standard Deviation: 5
#   Sample Size: 100
#   Probability: 95%

import numpy as np
import scipy.stats as stats

mean = 65
std_dev = 5
n = 100
ci = stats.norm.interval(0.95, loc=mean, scale=std_dev / np.sqrt(n))
print(f"95% Confidence Interval: {ci}")
```

```python
# 3. Suppose the battery life of a premium smartwatch on a single charge is normally distributed
#   with a mean of 56 hours and a standard deviation of 4 hours.
#   Calculate the probability that a randomly selected smartwatch lasts between 48 and 60 hours.

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

mu, sigma = 56, 4
prob = stats.norm.cdf(60, loc=mu, scale=sigma) - stats.norm.cdf(48, loc=mu, scale=sigma)
print(f"Probability: {prob:.4f}")

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)
plt.plot(x, y, label='Normal Distribution')
x_fill = np.linspace(48, 60, 500)
plt.fill_between(x_fill, stats.norm.pdf(x_fill, loc=mu, scale=sigma), color='skyblue', alpha=0.5, label='Between 48 and 60 hours')
plt.title('Battery Life Distribution')
plt.legend()
plt.show()
```

```python
# 4. Suppose the scores on a technical certification exam for IT professionals are
#   normally distributed with a mean of 78 points and a standard deviation of 10 points.
#   To earn a designation, an employee must score in the top 15% of all test-takers.
#   Calculate the minimum score required to earn this designation.
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

mu, sigma = 78, 10
min_score = stats.norm.ppf(0.85, loc=mu, scale=sigma)
print(f"Minimum score for top 15%: {min_score:.2f}")

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)
plt.plot(x, y, label='Normal Distribution')
x_fill = np.linspace(min_score, mu + 4*sigma, 500)
plt.fill_between(x_fill, stats.norm.pdf(x_fill, loc=mu, scale=sigma), color='salmon', alpha=0.5, label='Top 15%')
plt.axvline(min_score, color='red', linestyle='--', label=f'Min Score: {min_score:.2f}')
plt.title('Certification Exam Scores Distribution')
plt.legend()
plt.show()
```

1. **Formula for Part 1:**
To find the critical z-score for a 95% confidence level (two-tailed) in a normal distribution, the inverse cumulative distribution function (percentile point function) is used:

$$z = \text{ppf}(1 - \alpha/2) = \text{ppf}(0.975) \approx 1.96$$



For a general value $x$, the z-score formula is:

$$z = \frac{x - \mu}{\sigma}$$
