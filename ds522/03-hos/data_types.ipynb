```python
# create structured data type
dt = np.dtype([('name', 'U20'), ('age', np.int64), ('points', np.float64)])
print(dt)

# <U20 is a Unicode string with a maximum length of 20 characters
# <i8 is a 64-bit integer
# <f8 is a 64-bit float

# apply it to ndarray object called student
student = np.array([('student_A', 20, 100),
                    ('student_B_has_a_very_long_name', 20, 98),
                    ('student_C', 21, 95)], dtype=dt)

student
```

### Primary Use Cases

Multi-dimensional arrays are essential for handling data structured in grids, tensors, or matrices. They form the backbone of:

* **Data Analysis & Statistics:** Organizing tabular data and time-series sets.
* **Machine Learning & Deep Learning:** Representing model weights, features, and image tensors.
* **Image Processing:** Manipulating pixels structured across height, width, and color channels.
* **Scientific Computing:** Solving complex linear algebra, physics simulations, and coordinate geometry.

### Why NumPy Outperforms Python Lists

The performance gap comes down to **how they store data in memory**:

* **NumPy (Contiguous Block):** Stores data in a single, uninterrupted block of memory with uniform data types. This allows the CPU to leverage **caching** (loading neighboring data ahead of time) and **vectorization** (performing SIMD operations directly in highly optimized C code).
* **Python Lists (Pointers):** Python nested lists are arrays of pointers scattered across memory. Every element is a separate Python object. To read the list, the CPU must constantly jump around memory to resolve these pointers (pointer chasing), which destroys cache efficiency and prevents hardware-level vectorization.
