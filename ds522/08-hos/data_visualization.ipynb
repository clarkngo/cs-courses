```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
# x is a NumPy array with 256 values ranging from -PI to +PI (included). The endpoint=True parameter tells the function to include the last value
x = np.linspace(-np.pi, np.pi, 256, endpoint=True)
c, s = np.cos(x), np.sin(x)
```

```python
plt.plot(x,c, '-c') # - means solid line and c means cyan color
plt.plot(x,s, '--m') # -- means dashed line and m means magenta color
```

```python
# Initialize a figure
fig = plt.figure()
# add axes to the figure
ax = fig.add_axes([0,0,1,1])

y=[1,4,9,16,25,36,49,64]
x1=[1,16,30,42,55,68,77,88] # AMZ
x2=[1,6,12,18,28,40,52,65] # MSFT

l1 = ax.plot(x1, y, 'bs-') # solid line with blue color and square marker
l2 = ax.plot(x2, y, 'ro--') # dahsed line with red color and circle marker
ax.legend(labels=('AMZ','MSFT'), loc='lower right')
ax.set_title('Stock Market')
ax.set_xlabel('Time')
ax.set_ylabel('USD')
plt.show()
```

```python
x = np.arange(0, np.pi*2, 0.05)
y = np.sin(x)
z = np.cos(x)

# Create a figure
fig2 = plt.figure()

# set up axes
axes1 = fig2.add_axes([0.1,0.1,0.8,0.8]) # main axes (blue)
axes2 = fig2.add_axes([0.55, 0.55, 0.3, 0.3]) # inserted axes (red)

# plotting
axes1.plot(x,y,'b')
axes2.plot(x,z,'r')
axes1.set_title('sine')
axes2.set_title('cosine')
plt.show()
```

```python
n = 12
x = np.arange(n)
y1 = (1-x/float(n))*np.random.uniform(0.5,1.0,n)
y2 = (1-x/float(n))*np.random.uniform(0.5,1.0,n)

plt.bar(x, +y1, facecolor='#9999ff',edgecolor=['green','red'])
plt.bar(x, -y2, facecolor='#ff9999',edgecolor=['white'])

for x,y in zip(x,y1):
    # text properties and layout
    plt.text(x+0.4, y+0.05,f"{y:.2g}", ha='center',va='center')

plt.ylim(-1.25, +1.25)
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = Axes3D(fig)
fig.add_axes(ax)

x = np.arange(-4,4,0.25)
y = np.arange(-4,4,0.25)

# Mesgrid returns coordinates matrices from coordinate vectors
x,y=np.meshgrid(x,y)
r=np.sqrt(x**2+y**2)
z=np.sin(r)

ax.plot_surface(x,y,z,rstride=1,cstride=1,cmap='winter')

plt.show()
```

```python
# load tips dataset from seaborn
df = sb.load_dataset('tips')
sb.relplot(x='total_bill',y='tip',data=df, kind='scatter')

# scatter is default. Add the parameter kind='line' to change line plot
```

```python
sb.relplot(x='total_bill',y='tip', hue=df.day.tolist(),data=df)
# hue separates the data by day
```

```python
sb.set(style='whitegrid')
sb.relplot(x='total_bill', y='tip', hue=df.day.tolist(),palette='Greens',data=df)
# six default themes: [deep, muted, paste, bright, dark, and colorblind]
```

```python
sb.palplot(sb.color_palette('muted'))
sb.palplot(sb.color_palette('husl',8))
sb.palplot(sb.color_palette('Blues'))
```

```python
sb.catplot(x='day', y='total_bill', data=df, hue='day', kind='strip')
```

```python
# Load flights dataset and convert to long-form
flights_long = sb.load_dataset('flights')
flights = flights_long.pivot(index='month', columns='year',values='passengers')

# Draw a heatmap with the numeric values in each cell
f, ax = plt.subplots(figsize=(12,9))
sb.heatmap(flights, annot=True, fmt='d', linewidths=.5, ax=ax)
```
