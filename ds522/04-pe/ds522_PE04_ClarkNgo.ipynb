```python
import pandas as pd

# =====================================================================
# 1. Create a pandas dataframe called sales_df
# =====================================================================
# Loading the data from sales.csv
sales_df = pd.read_csv('sales.csv')
```

```python


# =====================================================================
# 2. Inspect the Data
# =====================================================================
print("--- 2. Inspecting Data ---")
# View the first few rows of the DataFrame
print("\nFirst 5 rows of the DataFrame:")
print(sales_df.head())

# Get the summary information of the DataFrame
print("\nDataFrame Info:")
sales_df.info()

# Get the summary statistics for the numerical columns
print("\nSummary Statistics:")
print(sales_df.describe())

```

```python

# =====================================================================
# 3. Basic Operations
# =====================================================================
print("\n--- 3. Basic Operations ---")
# Convert the Date column to a datetime object
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# Add a new column 'Total Sales' (Quantity * Price)
sales_df['Total Sales'] = sales_df['Quantity'] * sales_df['Price']
print("'Total Sales' column added and 'Date' converted successfully.")

```

```python

# =====================================================================
# 4. Filtering Data
# =====================================================================
print("\n--- 4. Filtering Data ---")
# Filter to get all sales that occurred in the month of July
# (.dt.month == 7 checks for July)
july_sales = sales_df[sales_df['Date'].dt.month == 7]
print(f"\nSales in July (Total rows: {len(july_sales)}):")
print(july_sales.head())

# Get all sales where the Quantity sold is greater than 10
large_quantity_sales = sales_df[sales_df['Quantity'] > 10]
print(f"\nSales with Quantity > 10 (Total rows: {len(large_quantity_sales)}):")
print(large_quantity_sales.head())

```

```python

# =====================================================================
# 5. Unique and Statistics
# =====================================================================
print("\n--- 5. Unique and Statistics ---")
# Get unique store names
unique_stores = sales_df['Store'].unique()
print(f"Unique Stores: {unique_stores}")

# Compute the total sales per store
total_sales_per_store = sales_df.groupby('Store')['Total Sales'].sum()
print("\nTotal Sales per Store:")
print(total_sales_per_store)

# Get unique products
unique_products = sales_df['Product'].unique()
print(f"\nUnique Products: {unique_products}")

# Calculate the average price of each product
avg_price_per_product = sales_df.groupby('Product')['Price'].mean()
print("\nAverage Price per Product:")
print(avg_price_per_product)

```

```python

# =====================================================================
# 6. Sorting and Ranking
# =====================================================================
print("\n--- 6. Sorting and Ranking ---")
# Sort the DataFrame by Total Sales in descending order
sorted_df = sales_df.sort_values(by='Total Sales', ascending=False)
print("\nDataFrame sorted by Total Sales (Descending):")
print(sorted_df.head())

# Rank the stores based on Total Sales
# First, get total sales per store, then rank them (1 = highest sales)
store_ranking = total_sales_per_store.rank(ascending=False, method='min')
print("\nStore Ranking based on Total Sales (1 is top performer):")
print(store_ranking)

```

```python

# =====================================================================
# 7. Apply Custom Functions
# =====================================================================
# Note: Since item #7 was cut off in the image, here is a common example 
# of applying a custom function (e.g., categorizing sales performance)
print("\n--- 7. Apply Custom Functions (Example) ---")

def categorize_sale(sales_value):
    if sales_value > 500:
        return 'High'
    elif sales_value > 100:
        return 'Medium'
    else:
        return 'Low'

sales_df['Sales_Category'] = sales_df['Total Sales'].apply(categorize_sale)
print(sales_df[['Total Sales', 'Sales_Category']].head())
```
